```python
import matplotlib.pyplot as plt

y1 = []
y2 = []

x = range(-100, 100, 10)

for i in x:
    y1.append(i**2)

for i in x:
    y2.append(-i**2)

plt.plot(x, y1)
plt.plot(x, y2)

plt.xlabel("x")
plt.ylabel("y")

plt.ylim(-2000, 2000)

plt.axhline(0)
plt.axvline(0)

plt.savefig("quad.png")
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris(as_frame=True)
df = iris.frame

plt.subplot(2,3,1)
plt.plot(df['sepal length (cm)'], df['petal length (cm)'])

plt.subplot(2,3,2)
plt.plot(df['sepal length (cm)'], df['petal length (cm)'])

plt.subplot(2,3,3)
plt.scatter(df['sepal length (cm)'], df['petal length (cm)'], c='green')
plt.title('subplott')

plt.show()
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

iris=sns.load_dataset("iris")

setosa=iris[iris["species"]=="setosa"]

sns.kdeplot(x=setosa.petal_length, y=setosa.petal_width,fill=True)

plt.title("KDE Plot of Setosa Petal Dimensions")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.show()
```
