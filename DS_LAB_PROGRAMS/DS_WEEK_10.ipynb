```python
# ============================================================
# DS_WEEK10
# Week 10: Regression Models
# ============================================================

# ------------------------------------------------------------
# 10.1 Linear Regression
# ------------------------------------------------------------

import numpy as np
from sklearn.linear_model import LinearRegression

X = np.array([[1000], [1200], [1500], [1800], [2000]])
y = np.array([50, 55, 65, 80, 90])

model = LinearRegression()
model.fit(X, y)

test_size = np.array([[1700]])
predicted_price = model.predict(test_size)

print("House Sizes:")
print(X.flatten())

print("\nHouse Prices:")
print(y)

print("\nPredicted Price for 1700 sq.ft.:")
print(predicted_price)

print("\nSlope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
# ------------------------------------------------------------
# 10.2 Logistic Regression
# ------------------------------------------------------------

import numpy as np
from sklearn.linear_model import LogisticRegression

X = np.array([[40], [50], [60], [70], [80], [90]])
y = np.array([0, 0, 0, 1, 1, 1])

model = LogisticRegression()
model.fit(X, y)

test_price = np.array([[75]])

prediction = model.predict(test_price)
probability = model.predict_proba(test_price)

print("House Prices:")
print(X.flatten())

print("\nLabels:")
print(y)

print("\nPrediction for price 75:")
print(prediction)

print("\nPrediction Probabilities:")
print(probability)
```
